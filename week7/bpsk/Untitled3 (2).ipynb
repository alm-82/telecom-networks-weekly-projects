```python
import numpy as np
import matplotlib.pyplot as plt
import random

print("╔══════════════════════════════════════════════════════════╗")
print("║               BPSK MODULATION SIMULATION                ║")
print("╚══════════════════════════════════════════════════════════╝")

class BPSKSimulator:
    def __init__(self, snr_db=10):
        self.snr_db = snr_db
        self.results = []
        
    def generate_bits(self, n_bits):
        return [random.randint(0, 1) for _ in range(n_bits)]
    
    def bpsk_modulate(self, bits):
        return np.array([1 if bit == 1 else -1 for bit in bits])
    
    def add_awgn_noise(self, signal, snr_db):
        snr_linear = 10 ** (snr_db / 10)
        signal_power = np.mean(signal ** 2)
        noise_power = signal_power / snr_linear
        noise = np.random.normal(0, np.sqrt(noise_power), len(signal))
        return signal + noise
    
    def bpsk_demodulate(self, signal):
        return [1 if s > 0 else 0 for s in signal]
    
    def calculate_ber(self, tx_bits, rx_bits):
        errors = sum(1 for t, r in zip(tx_bits, rx_bits) if t != r)
        return errors / len(tx_bits)
    
    def run_simulation(self, n_bits=100):
        print(f"\n{'━'*70}")
        print(f"🔧 BPSK SIMULATION - SNR = {self.snr_db} dB")
        print(f"{'━'*70}")
        
        tx_bits = self.generate_bits(n_bits)
        print(f"\n📤 TRANSMITTER:")
        print(f"Generated bits (first 20): {tx_bits[:20]}")
        
        modulated = self.bpsk_modulate(tx_bits)
        print(f"Modulated symbols: {modulated[:10]}...")
        
        noisy_signal = self.add_awgn_noise(modulated, self.snr_db)
        print(f"\n💥 CHANNEL (AWGN noise):")
        print(f"Noisy symbols: {noisy_signal[:10]}...")
        
        rx_bits = self.bpsk_demodulate(noisy_signal)
        print(f"\n📥 RECEIVER:")
        print(f"Demodulated bits (first 20): {rx_bits[:20]}")
        
        ber = self.calculate_ber(tx_bits, rx_bits)
        errors = sum(1 for t, r in zip(tx_bits, rx_bits) if t != r)
        
        print(f"\n🔍 RESULTS:")
        print(f"Total bits: {n_bits}")
        print(f"Errors: {errors}")
        print(f"BER: {ber:.6f} ({ber*100:.2f}%)")
        
        self.results.append({
            'snr': self.snr_db,
            'ber': ber,
            'errors': errors,
            'total': n_bits
        })
        
        return tx_bits, modulated, noisy_signal, rx_bits, ber
    
    def visualize_constellation(self, tx_signal, rx_signal):
        fig, axes = plt.subplots(1, 2, figsize=(12, 5))
        
        axes[0].scatter(tx_signal, np.zeros_like(tx_signal), color='blue', alpha=0.6)
        axes[0].axhline(y=0, color='k', linestyle='--', alpha=0.3)
        axes[0].axvline(x=0, color='k', linestyle='--', alpha=0.3)
        axes[0].set_xlim(-2, 2)
        axes[0].set_ylim(-1, 1)
        axes[0].set_title('Transmitter Constellation (BPSK)')
        axes[0].set_xlabel('In-phase (I)')
        axes[0].set_ylabel('Quadrature (Q)')
        axes[0].grid(True, alpha=0.3)
        
        axes[1].scatter(rx_signal, np.zeros_like(rx_signal), color='red', alpha=0.6)
        axes[1].axhline(y=0, color='k', linestyle='--', alpha=0.3)
        axes[1].axvline(x=0, color='k', linestyle='--', alpha=0.3)
        axes[1].set_xlim(-2, 2)
        axes[1].set_ylim(-1, 1)
        axes[1].set_title('Receiver Constellation (With Noise)')
        axes[1].set_xlabel('In-phase (I)')
        axes[1].grid(True, alpha=0.3)
        
        plt.tight_layout()
        plt.show()
    
    def visualize_waveforms(self, tx_bits, modulated, noisy_signal, rx_bits):
        fig, axes = plt.subplots(3, 1, figsize=(12, 8))
        
        time = np.arange(len(tx_bits))
        axes[0].step(time[:50], tx_bits[:50], where='post', linewidth=2)
        axes[0].set_title('Transmitter Bits')
        axes[0].set_xlabel('Time (bit index)')
        axes[0].set_ylabel('Bit Value')
        axes[0].set_ylim(-0.5, 1.5)
        axes[0].grid(True, alpha=0.3)
        
        axes[1].plot(time[:50], modulated[:50], 'b-', linewidth=2)
        axes[1].set_title('BPSK Modulated Signal')
        axes[1].set_xlabel('Time (bit index)')
        axes[1].set_ylabel('Amplitude')
        axes[1].grid(True, alpha=0.3)
        
        axes[2].plot(time[:50], noisy_signal[:50], 'r-', linewidth=2, alpha=0.7)
        axes[2].plot(time[:50], modulated[:50], 'b-', linewidth=1, alpha=0.5, label='Original')
        axes[2].set_title('Received Signal (With AWGN Noise)')
        axes[2].set_xlabel('Time (bit index)')
        axes[2].set_ylabel('Amplitude')
        axes[2].legend()
        axes[2].grid(True, alpha=0.3)
        
        plt.tight_layout()
        plt.show()
    
    def ber_vs_snr_simulation(self, snr_range=range(0, 16, 2), n_bits=10000):
        print(f"\n{'━'*70}")
        print(f"📊 BER vs SNR ANALYSIS")
        print(f"{'━'*70}")
        
        ber_results = []
        
        for snr in snr_range:
            self.snr_db = snr
            _, _, _, _, ber = self.run_simulation(n_bits)
            ber_results.append(ber)
            print(f"SNR = {snr:2d} dB → BER = {ber:.6f}")
        
        plt.figure(figsize=(10, 6))
        plt.semilogy(list(snr_range), ber_results, 'bo-', linewidth=2, markersize=8)
        plt.xlabel('SNR (dB)')
        plt.ylabel('Bit Error Rate (BER)')
        plt.title('BPSK Performance: BER vs SNR')
        plt.grid(True, which="both", ls="--", alpha=0.3)
        plt.yscale('log')
        
        snr_linear = 10 ** (np.array(snr_range) / 10)
        theoretical_ber = 0.5 * (1 - np.sqrt(snr_linear / (1 + snr_linear)))
        plt.semilogy(snr_range, theoretical_ber, 'r--', linewidth=2, label='Theoretical BPSK')
        
        plt.legend()
        plt.tight_layout()
        plt.show()

print("\n🔧 BPSK Modulation/Demodulation Simulation")
print("BPSK: 0 → -1 (180° phase), 1 → +1 (0° phase)")

simulator = BPSKSimulator(snr_db=5)

tx_bits, modulated, noisy_signal, rx_bits, ber = simulator.run_simulation(100)

print(f"\n{'━'*70}")
print("🌌 CONSTELLATION DIAGRAMS")
print(f"{'━'*70}")
simulator.visualize_constellation(modulated, noisy_signal)

print(f"\n{'━'*70}")
print("📈 SIGNAL WAVEFORMS")
print(f"{'━'*70}")
simulator.visualize_waveforms(tx_bits, modulated, noisy_signal, rx_bits)

simulator.ber_vs_snr_simulation(snr_range=range(0, 15, 3), n_bits=5000)

print(f"\n{'━'*70}")
print("💡 BPSK KEY CONCEPTS")
print(f"{'━'*70}")

print("""
📡 BPSK (Binary Phase Shift Keying):
• 0 bit → Phase shift of 180° (Amplitude = -1)
• 1 bit → Phase shift of 0° (Amplitude = +1)

🎯 Advantages:
• Simple implementation
• Robust against noise
• Constant envelope

⚠️ Disadvantages:
• Low spectral efficiency
• Sensitive to phase noise

📊 BER Formula for BPSK (AWGN channel):
BER = 0.5 × erfc(√(E_b/N_0))

Where:
• erfc = complementary error function
• E_b/N_0 = Energy per bit to noise power ratio

🔬 Applications:
• Satellite communications
• Deep space communications
• RFID systems
• Some WiFi standards
""")
```
